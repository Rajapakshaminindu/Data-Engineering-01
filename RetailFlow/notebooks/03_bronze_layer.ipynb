# RetailFlow - Bronze Layer
Ingests raw CSV datasets into Bronze Parquet format for fast, columnar analytical processing.

```python
from pathlib import Path
import pandas as pd
import re

# Dynamically resolve project root from any folder or subfolder
current = Path.cwd()
project_root = None
for candidate in [current, current / 'RetailFlow', *current.parents]:
    if (candidate / 'data' / 'raw').exists():
        project_root = candidate
        break
    if (candidate / 'RetailFlow' / 'data' / 'raw').exists():
        project_root = candidate / 'RetailFlow'
        break
if project_root is None:
    raise FileNotFoundError('RetailFlow root directory with data/raw not found.')

raw_path = project_root / 'data' / 'raw'
bronze_path = project_root / 'data' / 'bronze'
bronze_path.mkdir(parents=True, exist_ok=True)

raw_files = sorted(raw_path.glob('*.csv'))
print(f'Ingesting {len(raw_files)} CSV files into Bronze Layer...')

for file_path in raw_files:
    df = pd.read_csv(file_path)
    output_name = file_path.stem + '.parquet'
    df.to_parquet(bronze_path / output_name, index=False)
    print(f'  Saved: {output_name:<42} | Rows: {len(df):>10,}')

print('\nBronze layer completed successfully!')
```
